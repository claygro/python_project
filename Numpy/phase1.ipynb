## Numpy array and basics

```python
import numpy as np
```

### Creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("1D array: ", arr_1d)

arr_2d = np.array([[1, 2, 3],[4, 5, 6]])
print("2D array: ", arr_2d)
```

### List va numpy array

```python
py_list = [1, 2, 3]
print("Python list multiplication: ", py_list * 2)



np_array = np.array([1, 2, 3]) # element wise multiplication
print("numpy array multiplication: ", np_array * 2)


import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n numpy operation time: ", time.time() - start)

```

### Creating array from scratch

```python
zeros = np.zeros((3, 4)) # 3 row 4 column
print("zeros array: \n", zeros)

ones = np.ones((2, 3)) # 2 row 3 column
print("One array: \n", ones)

full = np.full((2, 2), 7) # 2 row 2 column
print("Full array: \n", full)

random = np.random.random([2, 3])
print("random array: \n", random)

sequence = np.arange(0, 10, 2) # start, stop, step
print("Sequence array: ", sequence)
```

### Vector, Matices and Tensor

```python
vector = np.array([1, 2, 3])
print("Vector: ", vector)

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrices: ", matrix)

tensor = np.array([[[1, 2,], [3, 4]],
                   [[5, 6], [7, 8]]]) # in tensor, there is more dimension
print("Tensor: ", tensor)
```

### Array properties

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])

print("Array: ", arr)
print("Shape: ", arr.shape)
print("Dimension: ", arr.ndim)
print("Size: ", arr.size)
print("Datatype: ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12) # n-1 means 12-1 = 11
print("Original array: ", arr)

reshaped = arr.reshape((3, 4)) # 3 row 4 column
print("Reshape array: ", reshaped)

flattened = reshaped.flatten() #returns copy
print("\n Flat: ", flattened)
# return view instead of copy
raveled = reshaped.ravel() #returns original array
print("\n raveled array: ", raveled)

# transpose
transpose = reshaped.T
print("\n Transposed array: ", transpose)
```
